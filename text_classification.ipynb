```python
from transformers import pipeline
import torch
fiction_categories =["Fiction", "Nonfiction"]
```

```python
pipe = pipeline("zero-shot-classification",
                model="facebook/bart-large-mnli")
```

```python
sequence=movies.loc[movies["simple_category"] == "Fiction", "description"].reset_index(drop=True)[0]
```

```python
pipe(sequence, fiction_categories)
```

now we are getting the score based on the description whether the movie/tv is fiction or non-fiction

```python
import numpy as np
max_index = np.argmax(pipe(sequence, fiction_categories)["scores"])
max_lable =pipe(sequence, fiction_categories)["labels"][max_index]
max_lable
```

creating function for all the data

```python
def generate_predictions(sequence,categories):
    predictions =pipe(sequence, categories)
    max_index = np.argmax(predictions["scores"])
    max_lable = predictions["labels"][max_index]
    return max_lable
```

Evaluating the zero shot model on how accurate the prediction is by comparing to know lables

```python
from tqdm import tqdm

actual_cats =[]
predicted_cats =[]

for i in tqdm(range(0, 300)):
    sequence = movies.loc[movies["simple_category"] == "Fiction", "description"].reset_index(drop=True)[i]
    predicted_cats += [generate_predictions(sequence, fiction_categories)]
    actual_cats += ["Fiction"]
```

```python
for i in tqdm(range(0, 200)):
    sequence = movies.loc[movies["simple_category"] == "Nonfiction", "description"].reset_index(drop=True)[i]
    predicted_cats += [generate_predictions(sequence, fiction_categories)]
    actual_cats += ["Nonfiction"]
```

```python
predictions_df =pd.DataFrame({"actual_category":actual_cats,"predicted_category":predicted_cats})
```

```python
predictions_df
```

```python
predictions_df["correct_prediction"] = (
    np.where(predictions_df["actual_category"] == predictions_df["predicted_category"], 1, 0)
)
```

```python
predictions_df["correct_prediction"].sum() / len(predictions_df)
```

```python
show_id =[]
predicted_cats = []

missing_cats = movies.loc[movies["simple_category"].isna(), ["show_id", "description"]].reset_index(drop=True)
```

```python
for i in tqdm(range(0, len(missing_cats))):
    sequence = missing_cats["description"][i]
    predicted_cats += [generate_predictions(sequence, fiction_categories)]
    show_id += [missing_cats["show_id"][i]]
```

```python
missing_predictions_df = pd.DataFrame({"show_id1":show_id, "predicted_category": predicted_cats})
```

```python
missing_predictions_df
```

merging with actual data with missing

```python
movies = pd.merge(
    movies,
    missing_predictions_df,
    left_on="show_id",         # column in movies
    right_on="show_id1",       # column in missing_predictions_df
    how="left"
)
movies["simple_category"] = np.where(movies["simple_category"].isna(), movies["predicted_category"], movies["simple_category"])
movies = movies.drop(columns = ["predicted_category"])
```

```python
movies = movies.drop(columns = ["predicted_category_x", "predicted_category_y"])
```

```python
movies = pd.merge(movies, missing_predictions_df, on="show_id1", how="right")
```
