```python
def retrieve_semantic_recommendations(
        query: str,
        top_k: int=10,
) -> pd.DataFrame:
    recs = db_movies.similarity_search(query,k=50)

    movies_list = []

    for i in range(0, len(recs)):
        movies_list+= [int(recs[i].page_content.split()[0].strip('"'))]

    return movies[movies["show_id"].isin(movies_list)].head(top_k)
```
